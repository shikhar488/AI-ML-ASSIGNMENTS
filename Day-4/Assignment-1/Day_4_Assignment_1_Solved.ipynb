## Q9. Handle missing values in the numerical columns appropriately.

**Answer:**

```python
numeric_columns = df.select_dtypes(include="number").columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing numerical values after handling:")
print(df[numeric_columns].isna().sum())
```

## Q10. Identify and remove duplicate records.

**Answer:**

```python
duplicate_count = df.duplicated().sum()
print("Duplicate records:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

## Q11. Check the Age column for invalid values.

**Answer:**

```python
if "Age" in df.columns:
    print("Minimum Age:", df["Age"].min())
    print("Maximum Age:", df["Age"].max())
    print("Invalid age records:")
    display(df[(df["Age"] < 0) | (df["Age"] > 120)])
else:
    print("Age column not found.")
```

## Q12. Correct or handle patients whose age is outside a reasonable range.

**Answer:**

```python
if "Age" in df.columns:
    invalid_age = (df["Age"] < 0) | (df["Age"] > 120)
    print("Invalid ages before correction:", invalid_age.sum())

    # Treat invalid ages as missing, then fill with the median age.
    df.loc[invalid_age, "Age"] = np.nan
    df["Age"] = df["Age"].fillna(df["Age"].median())

    print("Invalid ages after correction:", ((df["Age"] < 0) | (df["Age"] > 120)).sum())
```

## Q13. Identify inconsistent values in categorical columns such as Gender and Department.

**Answer:**

```python
categorical_columns = df.select_dtypes(include=["object", "category"]).columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].dropna().astype(str).value_counts())
```

## Q14. Standardize these categorical values.

**Answer:**

```python
# Standardize string-based categorical columns by trimming whitespace
# and applying consistent capitalization.
for col in categorical_columns:
    df[col] = df[col].apply(
        lambda x: x.strip().title() if isinstance(x, str) else x
    )

print("Categorical values standardized.")
```

## Q15. Create Age_Group using the specified ranges.

**Answer:**

```python
bins = [-np.inf, 17, 40, 60, np.inf]
labels = ["Child", "Young Adult", "Middle Age", "Senior"]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)

display(df[["Age", "Age_Group"]].head())
```

## Q16. Separate Blood_Pressure into Systolic and Diastolic columns.

**Answer:**

```python
if "Blood_Pressure" in df.columns:
    bp = df["Blood_Pressure"].astype(str).str.extract(
        r"(\d+)\s*/\s*(\d+)"
    )
    df["Systolic"] = pd.to_numeric(bp[0], errors="coerce")
    df["Diastolic"] = pd.to_numeric(bp[1], errors="coerce")
else:
    print("Blood_Pressure column not found.")

display(df.head())
```

## Q17. Create a High_Risk column based on suitable clinical conditions.

**Answer:**

```python
# A patient is marked high-risk if any of the following are present:
# high glucose, high cholesterol, high heart rate, or age above 60.
# Thresholds are explicitly defined here so the rule is reproducible.
conditions = []

if "Glucose" in df.columns:
    conditions.append(df["Glucose"] > 140)
if "Cholesterol" in df.columns:
    conditions.append(df["Cholesterol"] > 240)
if "Heart_Rate" in df.columns:
    conditions.append(df["Heart_Rate"] > 100)
if "Age" in df.columns:
    conditions.append(df["Age"] > 60)

if conditions:
    df["High_Risk"] = np.logical_or.reduce(conditions)
else:
    df["High_Risk"] = False

display(df[["High_Risk"]].head())
```

## Q18. Create Cost_Per_Day using treatment cost and length of stay.

**Answer:**

```python
cost_col = next((c for c in ["Treatment_Cost", "Treatment Cost", "TreatmentCost"] if c in df.columns), None)
stay_col = next((c for c in ["Length_of_Stay", "Length of Stay", "LengthOfStay"] if c in df.columns), None)

if cost_col and stay_col:
    df["Cost_Per_Day"] = df[cost_col] / df[stay_col].replace(0, np.nan)
    df["Cost_Per_Day"] = df["Cost_Per_Day"].fillna(0)
    display(df[[cost_col, stay_col, "Cost_Per_Day"]].head())
else:
    print("Treatment cost or length-of-stay column was not found.")
```

## Q24. Find the department with the highest average treatment cost.

## Q25. Calculate the percentage of patients who were readmitted.

**Answer:**

```python
readmit_col = next((c for c in ["Readmitted", "Readmission", "Readmitted_Status"] if c in df.columns), None)

if readmit_col:
    values = df[readmit_col].astype(str).str.strip().str.lower()
    yes_values = {"yes", "y", "1", "true", "readmitted"}
    readmitted_percentage = values.isin(yes_values).mean() * 100
    print("Readmitted percentage:", readmitted_percentage)
else:
    print("Readmission column was not found.")
```

## Q26. Compare readmission rates across departments.

**Answer:**

```python
readmit_col = next((c for c in ["Readmitted", "Readmission", "Readmitted_Status"] if c in df.columns), None)

if readmit_col:
    temp = df.copy()
    values = temp[readmit_col].astype(str).str.strip().str.lower()
    temp["_readmitted_flag"] = values.isin({"yes", "y", "1", "true", "readmitted"})
    rates = temp.groupby("Department")["_readmitted_flag"].mean() * 100
    print(rates.sort_values(ascending=False))
else:
    print("Readmission column was not found.")
```
