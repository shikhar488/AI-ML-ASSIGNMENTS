**Answer:**

```python
df = pd.read_csv("ecommerce_orders_raw.csv")
print("Dataset loaded successfully.")
display(df.head())
```

## Q2. Display its structure and dimensions.

**Answer:**

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nStructure:")
display(df.info())
```

## Q3. Identify the data types of all columns.

**Answer:**

```python
print(df.dtypes)
```

## Q4. Identify missing values.

**Answer:**

```python
print(df.isna().sum())
```

## Q5. Identify duplicate orders.

**Answer:**

```python
print("Number of duplicate rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)])
```

## Q6. Examine the unique values in City, Product Category, and Payment Method.

**Answer:**

```python
for col in ["City", "Product_Category", "Payment_Method"]:
    if col in df.columns:
        print(f"\n{col}:")
        print(df[col].dropna().unique())
```

## Q7. Handle missing ratings.

**Answer:**

```python
if "Rating" in df.columns:
    df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
    df["Rating"] = df["Rating"].fillna(df["Rating"].median())
    print("Missing ratings after handling:", df["Rating"].isna().sum())
```

## Q8. Standardize city names.

**Answer:**

```python
if "City" in df.columns:
    df["City"] = df["City"].astype("string").str.strip().str.title()
    print(df["City"].unique())
```

## Q9. Identify invalid quantities.

**Answer:**

```python
if "Quantity" in df.columns:
    df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
    invalid_quantity = df["Quantity"].isna() | (df["Quantity"] <= 0)
    print("Invalid quantity records:", invalid_quantity.sum())
    display(df[invalid_quantity])
```

## Q10. Handle duplicate orders.

**Answer:**

```python
duplicate_count = df.duplicated().sum()
df = df.drop_duplicates().reset_index(drop=True)
print("Duplicates removed:", duplicate_count)
print("Rows after removing duplicates:", len(df))
```

## Q11. Convert Order_Date into an appropriate datetime format.

**Answer:**

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
print(df["Order_Date"].dtype)
display(df[["Order_Date"]].head())
```

## Q12. Create Gross_Amount = Quantity × Unit_Price.

**Answer:**

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")
df["Gross_Amount"] = df["Quantity"] * df["Unit_Price"]
display(df[["Quantity", "Unit_Price", "Gross_Amount"]].head())
```

## Q13. Create Discount_Amount = Gross_Amount × Discount / 100.

**Answer:**

```python
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce").fillna(0)
df["Discount_Amount"] = df["Gross_Amount"] * df["Discount"] / 100
display(df[["Gross_Amount", "Discount", "Discount_Amount"]].head())
```

## Q14. Create Net_Amount = Gross_Amount - Discount_Amount.

**Answer:**

```python
df["Net_Amount"] = df["Gross_Amount"] - df["Discount_Amount"]
display(df[["Gross_Amount", "Discount_Amount", "Net_Amount"]].head())
```

## Q15. Create Rating_Category: 1–2 Poor, 3 Average, 4 Good, 5 Excellent.

**Answer:**

```python
def rating_category(rating):
    if rating in [1, 2]:
        return "Poor"
    elif rating == 3:
        return "Average"
    elif rating == 4:
        return "Good"
    elif rating == 5:
        return "Excellent"
    return "Unknown"

df["Rating_Category"] = df["Rating"].apply(rating_category)
display(df[["Rating", "Rating_Category"]].head())
```

## Q16. Extract Year, Month, Day, and Day of Week from Order_Date.

**Answer:**

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_of_Week"] = df["Order_Date"].dt.day_name()

display(df[["Order_Date", "Year", "Month", "Day", "Day_of_Week"]].head())
```

## Q17. Calculate total revenue.

**Answer:**

```python
total_revenue = df["Net_Amount"].sum()
print("Total Revenue:", total_revenue)
```

## Q18. Calculate total number of orders.

**Answer:**

```python
total_orders = df["Order_ID"].nunique() if "Order_ID" in df.columns else len(df)
print("Total Number of Orders:", total_orders)
```

## Q19. Find the average order value.

**Answer:**

```python
order_values = df.groupby("Order_ID")["Net_Amount"].sum()
average_order_value = order_values.mean()
print("Average Order Value:", average_order_value)
```

## Q20. Find total revenue by product category.

**Answer:**

```python
revenue_by_category = df.groupby("Product_Category")["Net_Amount"].sum().sort_values(ascending=False)
print(revenue_by_category)
```

## Q21. Find the top 10 products by revenue.

**Answer:**

```python
top_10_products = (
    df.groupby("Product")["Net_Amount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
print(top_10_products)
```

## Q22. Find the city generating the highest revenue.

**Answer:**

```python
city_revenue = df.groupby("City")["Net_Amount"].sum().sort_values(ascending=False)
print("City with highest revenue:", city_revenue.idxmax())
print("Revenue:", city_revenue.max())
```

## Q23. Find the most frequently used payment method.

**Answer:**

```python
payment_counts = df["Payment_Method"].value_counts()
print(payment_counts)
print("Most frequently used payment method:", payment_counts.idxmax())
```

## Q24. Find the average rating for each product category.

**Answer:**

```python
average_rating_by_category = df.groupby("Product_Category")["Rating"].mean().sort_values(ascending=False)
print(average_rating_by_category)
```

## Q25. Calculate monthly revenue.

**Answer:**

```python
monthly_revenue = (
    df.groupby(["Year", "Month"])["Net_Amount"]
      .sum()
      .sort_index()
)
print(monthly_revenue)
```

## Q26. Identify the highest-revenue month.

**Answer:**

```python
highest_month = monthly_revenue.idxmax()
highest_month_revenue = monthly_revenue.max()

print("Highest-revenue month:", highest_month)
print("Revenue:", highest_month_revenue)
```

## Q27. Find the top 10 customers by total spending.

**Answer:**

```python
top_10_customers = (
    df.groupby("Customer_ID")["Net_Amount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
print(top_10_customers)
```

## Q28. Compare revenue generated by different cities.

**Answer:**

```python
city_revenue = df.groupby("City")["Net_Amount"].sum().sort_values(ascending=False)
print(city_revenue)
```

## Q29. Find which product category has the highest average order value.

**Answer:**

```python
category_order_values = (
    df.groupby(["Order_ID", "Product_Category"])["Net_Amount"]
      .sum()
      .reset_index()
)

category_aov = (
    category_order_values.groupby("Product_Category")["Net_Amount"]
    .mean()
    .sort_values(ascending=False)
)

print(category_aov)
print("Category with highest average order value:", category_aov.idxmax())
print("Average Order Value:", category_aov.max())
```
