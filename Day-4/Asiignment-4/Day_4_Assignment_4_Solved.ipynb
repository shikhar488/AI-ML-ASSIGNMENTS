# Day 4 Assignment 4 — Customer, Product & Order Analysis

**Datasets:** `customers.csv`, `products.csv`, `orders.csv`

```python
import pandas as pd
import numpy as np

customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")

display(customers.head())
display(products.head())
display(orders.head())
```

## Q1. Load all three CSV files.

**Answer:**

```python
print("Customers:", customers.shape)
print("Products:", products.shape)
print("Orders:", orders.shape)
```

## Q2. Display the dimensions of each DataFrame.

**Answer:**

```python
print("customers:", customers.shape)
print("products:", products.shape)
print("orders:", orders.shape)
```

## Q3. Identify the columns in each dataset.

**Answer:**

```python
for name, frame in [("Customers", customers), ("Products", products), ("Orders", orders)]:
    print(f"\n{name} columns:")
    print(frame.columns.tolist())
```

## Q4. Identify the common columns between datasets.

**Answer:**

```python
print("Customers ∩ Orders:", sorted(set(customers.columns) & set(orders.columns)))
print("Products ∩ Orders:", sorted(set(products.columns) & set(orders.columns)))
print("Customers ∩ Products:", sorted(set(customers.columns) & set(products.columns)))
print("All three:", sorted(set(customers.columns) & set(products.columns) & set(orders.columns)))
```

## Q5. Determine how the three datasets are related.

**Answer:**

```python
print("Orders links to Customers through Customer_ID.")
print("Orders links to Products through Product_ID.")
print("Each order therefore combines customer details and product details.")
```

## Q6. Check each DataFrame for missing values.

**Answer:**

```python
for name, frame in [("Customers", customers), ("Products", products), ("Orders", orders)]:
    print(f"\n{name} missing values:")
    print(frame.isna().sum())
```

## Q7. Check for duplicate records.

**Answer:**

```python
for name, frame in [("Customers", customers), ("Products", products), ("Orders", orders)]:
    print(f"{name} duplicate rows:", frame.duplicated().sum())
```

## Q8. Check the validity of Customer_ID and Product_ID.

**Answer:**

```python
print("Customer_ID unique:", customers["Customer_ID"].is_unique)
print("Product_ID unique:", products["Product_ID"].is_unique)
print("Customer_ID pattern valid:", customers["Customer_ID"].astype(str).str.fullmatch(r"C\d{5}").all())
print("Product_ID pattern valid:", products["Product_ID"].astype(str).str.fullmatch(r"P\d{4}").all())
```

## Q9. Convert Order_Date into datetime.

**Answer:**

```python
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"], errors="coerce")
print("Order_Date dtype:", orders["Order_Date"].dtype)
print("Invalid dates:", orders["Order_Date"].isna().sum())
```

## Q10. Check whether every order corresponds to a valid customer.

**Answer:**

```python
invalid_customers = (~orders["Customer_ID"].isin(customers["Customer_ID"])).sum()
print("Orders with invalid Customer_ID:", invalid_customers)
print("Every order has a valid customer:", invalid_customers == 0)
```

## Q11. Check whether every order corresponds to a valid product.

**Answer:**

```python
invalid_products = (~orders["Product_ID"].isin(products["Product_ID"])).sum()
print("Orders with invalid Product_ID:", invalid_products)
print("Every order has a valid product:", invalid_products == 0)
```

## Q12. Merge orders with customers.

**Answer:**

```python
orders_customers = orders.merge(customers, on="Customer_ID", how="left", validate="many_to_one")
print(orders_customers.shape)
display(orders_customers.head())
```

## Q13. Merge the resulting DataFrame with products.

**Answer:**

```python
consolidated = orders_customers.merge(products, on="Product_ID", how="left", validate="many_to_one")
print(consolidated.shape)
display(consolidated.head())
```

## Q14. Create a final consolidated DataFrame.

**Answer:**

```python
final_df = consolidated[[
    "Order_ID","Customer_ID","Customer_Name","City","Age","Gender",
    "Product_ID","Product_Name","Category","Price","Quantity","Order_Date"
]].copy()
display(final_df.head())
print("Final dimensions:", final_df.shape)
```

## Q15. Create Order_Value = Price × Quantity.

**Answer:**

```python
final_df["Order_Value"] = final_df["Price"] * final_df["Quantity"]
display(final_df[["Order_ID","Price","Quantity","Order_Value"]].head())
```

## Q16. Calculate total revenue.

**Answer:**

```python
total_revenue = final_df["Order_Value"].sum()
print(f"Total revenue: ₹{total_revenue:,.2f}")
```

## Q17. Calculate average order value.

**Answer:**

```python
average_order_value = final_df["Order_Value"].mean()
print(f"Average order value: ₹{average_order_value:,.2f}")
```

## Q18. Find total revenue by product category.

**Answer:**

```python
revenue_by_category = final_df.groupby("Category")["Order_Value"].sum().sort_values(ascending=False)
display(revenue_by_category.to_frame("Total_Revenue"))
```

## Q19. Find total revenue by city.

**Answer:**

```python
revenue_by_city = final_df.groupby("City")["Order_Value"].sum().sort_values(ascending=False)
display(revenue_by_city.to_frame("Total_Revenue"))
```

## Q20. Find the top 10 customers by spending.

**Answer:**

```python
top10_customers = final_df.groupby(["Customer_ID","Customer_Name"])["Order_Value"].sum().sort_values(ascending=False).head(10)
display(top10_customers.to_frame("Total_Spending"))
```

## Q21. Find the top 10 products by revenue.

**Answer:**

```python
top10_products = final_df.groupby(["Product_ID","Product_Name"])["Order_Value"].sum().sort_values(ascending=False).head(10)
display(top10_products.to_frame("Total_Revenue"))
```

## Q22. Find the most frequently purchased product.

**Answer:**

```python
product_frequency = final_df.groupby(["Product_ID","Product_Name"])["Quantity"].sum().sort_values(ascending=False)
print("Most frequently purchased product:")
display(product_frequency.head(1).to_frame("Total_Quantity"))
print("\nTop 10 by quantity:")
display(product_frequency.head(10).to_frame("Total_Quantity"))
```

## Q23. Find the city with the highest number of customers.

**Answer:**

```python
customer_count_by_city = customers["City"].value_counts()
display(customer_count_by_city.to_frame("Customer_Count"))
print("City with highest number of customers:", customer_count_by_city.idxmax())
```

## Q24. Find the city generating the highest revenue.

**Answer:**

```python
print("City with highest revenue:", revenue_by_city.idxmax())
print(f"Revenue: ₹{revenue_by_city.max():,.2f}")
```

## Q25. Divide customers into age groups.

**Answer:**

```python
customers["Age_Group"] = pd.cut(
    customers["Age"], bins=[17,30,45,60,np.inf],
    labels=["18–30","31–45","46–60","61+"]
)
print(customers["Age_Group"].value_counts().sort_index())
```

## Q26. Calculate revenue generated by each age group.

**Answer:**

```python
final_df["Age_Group"] = pd.cut(
    final_df["Age"], bins=[17,30,45,60,np.inf],
    labels=["18–30","31–45","46–60","61+"]
)
revenue_by_age_group = final_df.groupby("Age_Group", observed=False)["Order_Value"].sum()
display(revenue_by_age_group.to_frame("Revenue"))
```

## Q27. Compare spending between male and female customers.

**Answer:**

```python
gender_spending = final_df.groupby("Gender")["Order_Value"].agg(
    Total_Spending="sum",
    Average_Order_Value="mean"
).sort_values("Total_Spending", ascending=False)
display(gender_spending)
print("Higher total spending:", gender_spending["Total_Spending"].idxmax())
```

## Q28. Identify high-value customers.

**Answer:**

```python
# High-value customers are defined here as customers whose spending is above the median customer spending.
customer_spending = final_df.groupby(["Customer_ID","Customer_Name"])["Order_Value"].sum().sort_values(ascending=False)
high_value_threshold = customer_spending.median()
high_value_customers = customer_spending[customer_spending > high_value_threshold]
print(f"High-value threshold (median spending): ₹{high_value_threshold:,.2f}")
print("Number of high-value customers:", len(high_value_customers))
display(high_value_customers.head(10).to_frame("Total_Spending"))
```

## Q29. Determine the percentage of customers who have placed more than one order.

**Answer:**

```python
orders_per_customer = orders["Customer_ID"].value_counts()
repeat_customers = (orders_per_customer > 1).sum()
percentage_repeat = repeat_customers / len(customers) * 100
print("Customers with more than one order:", repeat_customers)
print(f"Percentage of customers with more than one order: {percentage_repeat:.2f}%")
```

## Q30. Create a customer spending ranking.

**Answer:**

```python
customer_ranking = final_df.groupby(["Customer_ID","Customer_Name"])["Order_Value"].sum().sort_values(
    ascending=False
).reset_index(name="Total_Spending")
customer_ranking["Rank"] = customer_ranking["Total_Spending"].rank(method="min", ascending=False).astype(int)
customer_ranking = customer_ranking[["Rank","Customer_ID","Customer_Name","Total_Spending"]]
display(customer_ranking.head(20))
```

## Q31. Identify the top 10% of customers based on total spending.

**Answer:**

```python
n_top_10 = int(np.ceil(len(customer_ranking) * 0.10))
top_10_percent_customers = customer_ranking.head(n_top_10).copy()
print("Number of customers in top 10%:", n_top_10)
print(f"Spending by top 10%: ₹{top_10_percent_customers['Total_Spending'].sum():,.2f}")
display(top_10_percent_customers.head(20))
```

## Q32. Find the most profitable product category for each city.

**Answer:**

```python
city_category_revenue = final_df.groupby(["City","Category"])["Order_Value"].sum().reset_index()
idx = city_category_revenue.groupby("City")["Order_Value"].idxmax()
most_profitable_category_by_city = city_category_revenue.loc[idx].sort_values("City").reset_index(drop=True)
display(most_profitable_category_by_city)
```

## Q33. Find the most popular product in each category.

**Answer:**

```python
category_product_quantity = final_df.groupby(["Category","Product_Name"])["Quantity"].sum().reset_index()
idx = category_product_quantity.groupby("Category")["Quantity"].idxmax()
most_popular_product = category_product_quantity.loc[idx].sort_values("Category").reset_index(drop=True)
display(most_popular_product)
```

## Q34. Calculate monthly revenue.

**Answer:**

```python
monthly_revenue = final_df.set_index("Order_Date").resample("MS")["Order_Value"].sum()
display(monthly_revenue.to_frame("Monthly_Revenue"))
```

## Q35. Identify the month with the highest revenue.

**Answer:**

```python
best_month = monthly_revenue.idxmax()
best_month_revenue = monthly_revenue.max()
print("Month with highest revenue:", best_month.strftime("%B %Y"))
print(f"Highest monthly revenue: ₹{best_month_revenue:,.2f}")
```
