### Q21. Write a Python function called `calculate_average()` that accepts a list of numbers and returns their average. Do not use NumPy.

```python
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0

    total = 0
    for number in numbers:
        total += number

    return total / len(numbers)


numbers = [10, 20, 30, 40, 50]
print("Average:", calculate_average(numbers))
```

### Q22. Write a Python function called `find_maximum()` that accepts a list of numbers and returns the largest value without using Python's built-in `max()` function.

```python
def find_maximum(numbers):
    if len(numbers) == 0:
        return None

    largest = numbers[0]

    for number in numbers[1:]:
        if number > largest:
            largest = number

    return largest


numbers = [12, 45, 7, 89, 23]
print("Maximum:", find_maximum(numbers))
```

### Q24. Write a Python function called `count_even()` that accepts a list of integers and returns the number of even numbers in the list.

```python
def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count


numbers = [1, 2, 3, 4, 5, 6, 8]
print("Number of even numbers:", count_even(numbers))
```

### Q25. Write a modular Python program to calculate the accuracy of a classification model using `calculate_accuracy(y_true, y_pred)` and `display_result(accuracy)`. Do not use Scikit-learn's built-in accuracy function.

```python
def calculate_accuracy(y_true, y_pred):
    if len(y_true) != len(y_pred):
        raise ValueError("y_true and y_pred must have the same length.")

    if len(y_true) == 0:
        return 0

    correct = 0

    for actual, predicted in zip(y_true, y_pred):
        if actual == predicted:
            correct += 1

    return correct / len(y_true)


def display_result(accuracy):
    print(f"Accuracy: {accuracy:.2%}")


y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 0]

accuracy = calculate_accuracy(y_true, y_pred)
display_result(accuracy)
```

### Q26. Write a Python function that accepts a list of numbers and returns two values: mean and standard deviation.

```python
def calculate_mean_and_std(numbers):
    if len(numbers) == 0:
        return 0, 0

    mean = sum(numbers) / len(numbers)

    variance = 0
    for number in numbers:
        variance += (number - mean) ** 2

    variance = variance / len(numbers)
    standard_deviation = variance ** 0.5

    return mean, standard_deviation


numbers = [10, 20, 30, 40, 50]
mean, standard_deviation = calculate_mean_and_std(numbers)

print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
```

### Q27. Write a function called `normalize()` that accepts a list of numerical values and performs min-max normalization. Do not use Scikit-learn's preprocessing functions.

```python
def normalize(numbers):
    if len(numbers) == 0:
        return []

    x_min = min(numbers)
    x_max = max(numbers)

    if x_max == x_min:
        return [0.0 for _ in numbers]

    normalized_values = []

    for x in numbers:
        x_norm = (x - x_min) / (x_max - x_min)
        normalized_values.append(x_norm)

    return normalized_values


numbers = [10, 20, 30, 40, 50]
print("Original values:", numbers)
print("Normalized values:", normalize(numbers))
```

### Q28. Write a modular Python program for a simple machine-learning preprocessing pipeline using separate functions: `load_data()` → `remove_missing_values()` → `normalize_data()` → `display_data()`.

```python
def load_data():
    # Example data with missing values represented by None
    return [10, 20, None, 40, 50, None, 60]


def remove_missing_values(data):
    return [value for value in data if value is not None]


def normalize_data(data):
    if len(data) == 0:
        return []

    x_min = min(data)
    x_max = max(data)

    if x_max == x_min:
        return [0.0 for _ in data]

    normalized = []
    for value in data:
        normalized.append((value - x_min) / (x_max - x_min))

    return normalized


def display_data(data):
    print("Processed data:", data)


# Modular preprocessing pipeline
data = load_data()
data = remove_missing_values(data)
data = normalize_data(data)
display_data(data)
```

### Q29. Write a Python function to calculate the Euclidean distance between two points without using any built-in distance function.

```python
def euclidean_distance(point1, point2):
    if len(point1) != len(point2):
        raise ValueError("Both points must have the same number of dimensions.")

    squared_sum = 0

    for a, b in zip(point1, point2):
        squared_sum += (a - b) ** 2

    return squared_sum ** 0.5


point1 = (1, 2)
point2 = (4, 6)

print("Euclidean distance:", euclidean_distance(point1, point2))
```
