## Task 2 — From Vector to Matrix

**Answer:** Each row represents one time step and each column represents one sensor.


```python
temperature = [25, 26, 25, 27, 28]
pressure = [101, 102, 101, 103, 104]
humidity = [60, 62, 61, 63, 64]
vibration = [0.2, 0.3, 0.2, 0.4, 0.5]

X = np.array([temperature, pressure, humidity, vibration]).T

print("Tensor:")
print(X)
print("Shape:", X.shape)
print("Number of dimensions:", X.ndim)
print("First time step:", X[0])
print("Temperature readings:", X[:, 0])
print("Vibration readings:", X[:, 3])
print("Axis 0 = time, Axis 1 = sensors")
```

## Task 3 — Create a 3D Tensor

**Answer:** The tensor has dimensions **Machine × Time × Sensor = (10, 5, 4)**.


```python
# 10 machines × 5 time steps × 4 sensors
X = np.arange(10 * 5 * 4).reshape(10, 5, 4)

print("Shape:", X.shape)

# Machine 3, time step 2 (using Python zero-based indexing)
print("Temperature reading of machine 3 at time step 2:", X[2, 1, 0])

# All sensor readings of machine 5
print("All readings of machine 5:")
print(X[4])

# All machines at time step 1
print("All machines at time step 1:")
print(X[:, 0, :])

# Sensor 4 from all machines and all time steps
print("Sensor 4 readings:")
print(X[:, :, 3])
```

## Task 4 — Indexing a Tensor

Given `X = np.arange(24).reshape(2, 3, 4)`:

**Answers:**
1. Shape = **(2, 3, 4)**
2. `X[0]` represents the first block, with shape **(3, 4)**
3. `X[0,0]` represents the first row of the first block, with shape **(4,)**
4. `X[0,0,0]` represents one scalar element.


```python
X = np.arange(24).reshape(2, 3, 4)

print("X:")
print(X)
print("\nX[0]:")
print(X[0])
print("\nX[1]:")
print(X[1])
print("\nX[0, 0]:")
print(X[0, 0])
print("\nX[0, 0, 0]:")
print(X[0, 0, 0])

print("\nShape:", X.shape)
print("Shape of X[0]:", X[0].shape)
print("Shape of X[0,0]:", X[0,0].shape)
print("X[0,0,0] is a scalar:", np.isscalar(X[0,0,0]))
```

## Task 5 — Slicing

The required slices are:
- First block: `X[0]`
- First row from every block: `X[:, 0, :]`
- Third element from every row and block: `X[:, :, 2]`
- Last two elements: `X[:, :, -2:]`


```python
X = np.arange(60).reshape(3, 4, 5)

first_block = X[0]
first_row_every_block = X[:, 0, :]
third_element_every_row_block = X[:, :, 2]
last_two_elements = X[:, :, -2:]

print("First block:")
print(first_block)
print("Shape:", first_block.shape)

print("\nFirst row from every block:")
print(first_row_every_block)
print("Shape:", first_row_every_block.shape)

print("\nThird element from every row and block:")
print(third_element_every_row_block)
print("Shape:", third_element_every_row_block.shape)

print("\nLast two elements:")
print(last_two_elements)
print("Shape:", last_two_elements.shape)
```

## Task 8 — Reshape

**Answer:** Reshaping changes how the same elements are arranged across dimensions. It does not change the total number of elements, so every target shape must have a product of dimensions equal to 24.


```python
X = np.arange(24)

shapes = [(4, 6), (2, 3, 4), (6, 4), (24,)]

for shape in shapes:
    reshaped = X.reshape(shape)
    print("\nShape:", reshaped.shape)
    print("Dimensions:", reshaped.ndim)
    print(reshaped)

print("\nTotal elements:", X.size)
```

## Task 9 — Reshape vs Transpose

**Answer:** No. `reshape()` changes the arrangement of elements into a new shape, while `transpose()` changes the order of axes. They can produce different values even when the resulting shapes are the same.


```python
X = np.arange(24).reshape(2, 3, 4)

reshaped = X.reshape(3, 2, 4)
transposed = X.transpose(1, 0, 2)

print("Original:")
print(X)

print("\nReshape (3,2,4):")
print(reshaped)

print("\nTranspose (1,0,2):")
print(transposed)

print("\nAre they equal?", np.array_equal(reshaped, transposed))
```

## Task 10 — Broadcasting

**Answer:** A `(4,)` array is compatible with `(3,4)` because NumPy aligns dimensions from the right. The 4 elements are applied independently to each of the 3 rows.


```python
X = np.ones((3, 4))
b = np.array([10, 20, 30, 40])

print("X + b:")
print(X + b)

print("\nX * b:")
print(X * b)
```

## Task 12 — Tensor Aggregation

For `X.shape = (100, 20, 5)`:
- Mean across time → **(100, 5)**
- Mean across sensors → **(100, 20)**
- Mean across machines → **(20, 5)**
- Maximum across time → **(100, 5)**
- Standard deviation across time → **(100, 5)**


```python
X = np.random.randn(100, 20, 5)

mean_time = X.mean(axis=1)
mean_sensors = X.mean(axis=2)
mean_machines = X.mean(axis=0)
max_time = X.max(axis=1)
std_time = X.std(axis=1)

print("Mean across time:", mean_time.shape)
print("Mean across sensors:", mean_sensors.shape)
print("Mean across machines:", mean_machines.shape)
print("Maximum across time:", max_time.shape)
print("Standard deviation across time:", std_time.shape)
```

## Task 13 — Normalize a Tensor

**Answer:** Normalize each sensor using its mean and standard deviation calculated over machines and time steps.

Formula:

`X_normalized = (X - mean) / std`


```python
X = np.random.randn(100, 20, 5)

# Mean/std for each sensor: keep sensor axis and reduce machine/time axes
mean = X.mean(axis=(0, 1), keepdims=True)
std = X.std(axis=(0, 1), keepdims=True)

X_normalized = (X - mean) / std

print("Normalized shape:", X_normalized.shape)
print("Mean of each sensor after normalization:")
print(X_normalized.mean(axis=(0, 1)))
print("\nStandard deviation of each sensor after normalization:")
print(X_normalized.std(axis=(0, 1)))
```

## Task 16 — Practical Mini Challenge: Sensor Dataset

Create a tensor with shape **(500, 60, 8)** representing:

**Machines × Time × Sensors**


```python
X = np.random.randn(500, 60, 8)

# 1. Machine 10
machine_10 = X[9]

# 2. First 20 time steps
first_20_time_steps = X[:, :20, :]

# 3. Sensor 5 from all machines
sensor_5 = X[:, :, 4]

# 4. Average value of every sensor
average_each_sensor = X.mean(axis=(0, 1))

# 5. Average value at every time step
average_each_time_step = X.mean(axis=(0, 2))

# 6. Machine with highest average sensor reading
machine_average = X.mean(axis=(1, 2))
highest_machine = np.argmax(machine_average) + 1

# 7. Normalize each sensor
sensor_mean = X.mean(axis=(0, 1), keepdims=True)
sensor_std = X.std(axis=(0, 1), keepdims=True)
X_normalized = (X - sensor_mean) / sensor_std

# 8. Split into training and testing
X_train = X[:400]
X_test = X[400:]

print("Original shape:", X.shape)
print("Machine 10 shape:", machine_10.shape)
print("First 20 time steps shape:", first_20_time_steps.shape)
print("Sensor 5 shape:", sensor_5.shape)
print("Average of every sensor:", average_each_sensor)
print("Average at every time step shape:", average_each_time_step.shape)
print("Machine with highest average reading:", highest_machine)
print("Normalized shape:", X_normalized.shape)
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

## Task 17 — PyTorch

For `torch.arange(24).reshape(2,3,4)`:
- Shape = `(2,3,4)`
- Number of dimensions = `3`
- Number of elements = `24`

For `torch.randn(32,100,768)`, a common interpretation is:

**32 = batch size, 100 = sequence length/time steps, 768 = feature or embedding dimension.**


```python
import torch

X = torch.arange(24).reshape(2, 3, 4)

print(X)
print("Shape:", X.shape)
print("Number of dimensions:", X.ndim)
print("Number of elements:", X.numel())

X = torch.randn(32, 100, 768)

print("\nSecond tensor shape:", X.shape)
print("Possible interpretation:")
print("32  -> batch size")
print("100 -> sequence length / time steps")
print("768 -> feature or embedding dimension")
```
